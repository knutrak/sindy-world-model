## Setup

Demonstrations use the exp06 validation battles (1,000 close battles with intents switching every
1-4 s), the SINDYc model saved by `sindyc_02_world_model.ipynb`, the counterfactual sets and the
planner run.

```python
import pickle
import warnings

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from sklearn.ensemble import HistGradientBoostingRegressor
from sklearn.model_selection import cross_val_predict

from sindy_wm.envs.intents import INTENTS
from sindy_wm.models.intent_model import (
    DT,
    REACH_COLS,
    SHAPE,
    STATE,
    add_controls,
    add_rates,
    horizon_errors,
    make_batch,
    r2,
    share,
)
from sindy_wm.paths import DATA_DIR
from sindy_wm.states.geometry import MARINE_HP, REACH

warnings.filterwarnings("ignore")
pd.set_option("display.width", 200)
pd.set_option("display.max_colwidth", 120)
PROCESSED = DATA_DIR / "processed" / "exp06_intents"
LABELS = [f"{m}/{t}" for m, t in INTENTS]

val = add_rates(add_controls(pd.read_parquet(PROCESSED / "val.parquet")))
rows = val[(val["alive_blue"] > 0) & (val["alive_red"] > 0)]
P = pickle.load(open(PROCESSED / "closures.pkl", "rb"))
sparse = pickle.load(open(PROCESSED / "model_sparse.pkl", "rb"))
print(
    f"{val.index.get_level_values(0).nunique()} validation battles, {len(rows):,} rows with both sides alive"
)
```

## D1 · Losses follow units within reach (finding 1)

One-variable fits through the origin: how well does each candidate driver explain a side's
health loss rate? On the common scale (health as a share of 10 marines, units divided by 10),
a marine's maximum damage rate is 9.84 hp/s, i.e. 0.219 per unit.

```python
max_rate = 9.84 / MARINE_HP  # health share per s, per unit (common scale: /10 health, /10 units cancel)
table = {}
for side, opp in (("blue", "red"), ("red", "blue")):
    y = rows[f"d_health_{side}"].to_numpy()
    for driver in ("health", "alive", "reach"):
        x = rows[f"{driver}_{opp}"].to_numpy()
        k = np.dot(x, y) / np.dot(x, x)
        table[(f"{side}'s losses", f"opponent's {driver}")] = {"R²": r2(y, k * x), "k": -k}
fit = pd.DataFrame(table).T
fit.loc[(slice(None), "opponent's reach"), "share of a marine's max"] = (
    fit.loc[(slice(None), "opponent's reach"), "k"] / max_rate
)
fit.round(3)
```

Reach explains the losses far better than health or alive, and the coefficient is a
recognisable physical number: the share of a marine's maximum damage rate that is delivered.
(exp05 gave the same picture: `engagement_from_distance.ipynb`, section 9.)

## D2 · Reach follows distance, depth and width (finding 2)

The share of a side's units within reach of an enemy, against the distance between the team
centres. Dots: binned data. Line: the fitted closure, evaluated on the same rows.

```python
fig, axes = plt.subplots(1, 2, figsize=(11, 3.5), sharey=True)
bins = np.arange(2, 14, 0.25)
for ax, (side, opp) in zip(axes, (("blue", "red"), ("red", "blue")), strict=True):
    s_true = rows[f"reach_{side}"] / rows[f"alive_{side}"]
    s_pred = share(
        rows["distance"].to_numpy(),
        rows[f"depth_{side}"].to_numpy(),
        rows[f"depth_{opp}"].to_numpy(),
        rows[f"width_{side}"].to_numpy(),
        rows[f"width_{opp}"].to_numpy(),
        *P[f"share_{side}"],
    )
    by = (
        pd.DataFrame({"true": s_true, "closure": s_pred})
        .groupby(pd.cut(rows["distance"], bins), observed=True)
        .mean()
    )
    centres = [b.mid for b in by.index]
    ax.plot(centres, by["true"], "o", ms=3, label="data")
    ax.plot(centres, by["closure"], "-", label="closure")
    ax.axvline(REACH, color="k", ls=":", lw=1)
    rr = r2(rows[f"reach_{side}"].to_numpy(), rows[f"alive_{side}"].to_numpy() * s_pred)
    ax.set(xlabel="distance between team centres", title=f"{side}: reach R² {rr:.2f}")
axes[0].set_ylabel("share of units within reach")
axes[0].legend()
print("fitted switch c:", {s: round(P[f"share_{s}"][0], 2) for s in ("blue", "red")}, f"(game reach {REACH})")
```

## D3 · The SINDYc world model and its control terms (finding 3)

The sparse model from `sindyc_02_world_model.ipynb` (threshold 0.2, fitted on 6,000 training
battles). `Rb`, `Rr` = units within reach; `close*adv` etc. = closing gates per movement;
`wd` = withdraw.

```python
sparse.print()
```

How to read it:

- `(health_red)' = -0.185 Rb + 0.164 Rb*wd`: while withdrawing, Blue's fire drops by
  0.164 / 0.185 = 89%.
- `(distance)'`: about -6 when Blue advances (both sides walk at about 3), about -3 when Blue
  holds or withdraws (only Red walks); withdrawing in contact opens the distance (`flee*wd`).
- The alive exponent for Red rises with focus fire: the same health loss removes more units.

## D4 · Predicting ahead, and what limits it (findings 4, 5)

Start points every 1 s in the validation battles; the real intents are applied over the horizon.
Mean absolute error in health (share of 10 marines) and distance.

```python
B = make_batch(val)
steps = B["U"].shape[1]


def predict(model, **kw):
    return model.simulate(B["X0"], B["shape0"], B["U"], **kw)


results = {
    "persistence (nothing changes)": horizon_errors(np.repeat(B["X0"][:, None], steps + 1, 1), B),
    "last slope (trend of the last second)": horizon_errors(
        np.maximum(B["X0"][:, None] + B["slope"][:, None] * (np.arange(steps + 1) * DT)[None, :, None], 0), B
    ),
    "SINDYc sparse, self-contained": horizon_errors(predict(sparse), B),
    "SINDYc sparse + real distance, shape, alive": horizon_errors(
        predict(
            sparse, truth={"distance": B["Y"][:, :, 4], "shape": B["Yshape"], "alive": B["Y"][:, :, 2:4]}
        ),
        B,
    ),
    "SINDYc sparse + real reach (upper bound)": horizon_errors(
        predict(sparse, truth={"reach": B["Yreach"]}), B
    ),
}
pd.DataFrame(results).T.round(3)
```

For reference (not recomputed here): without control the physics model gets 0.079 at 4 s; a
generic degree-2 SINDYc gets 0.055 with 147 terms (finding 8); on exp05 the self-contained model
of `engagement_from_distance.ipynb` section 14 gets 0.040 against 0.065 for health only.

## D5 · Determinism and leverage: from one state, the intent changes the outcome (findings 6, 7)

Counterfactual set `cf`: a battle is played up to a drawn moment, then each of the six intents
for 4 s, then advance/nearest to the end. The six runs share everything up to the branch.

```python
def load_counterfactuals(prefix):
    tables = {i: pd.read_parquet(PROCESSED / f"{prefix}_{i[0]}_{i[1]}.parquet") for i in INTENTS}
    ref = tables[INTENTS[0]]
    records, starts = [], []
    for ep, g in ref.groupby(level=0):
        k = int(round(g["branch_at"].iloc[0] / DT))
        if k >= len(g) - 8 or g["alive_blue"].iloc[k] <= 0 or g["alive_red"].iloc[k] <= 0:
            continue
        rec = {}
        for label, i in zip(LABELS, INTENTS, strict=True):
            h = tables[i].loc[ep]
            assert (
                abs(h["health_blue"].iloc[k] - g["health_blue"].iloc[k]) < 1e-9
            )  # identical up to the branch
            rec[f"final {label}"] = h["health_blue"].iloc[-1] - h["health_red"].iloc[-1]
            rec[f"win {label}"] = bool(h["battle_won"].iloc[0])
        records.append(rec)
        starts.append(g.iloc[k][STATE + SHAPE + REACH_COLS])
    return pd.DataFrame(records), pd.DataFrame(starts).reset_index(drop=True)


cf, cf_start = load_counterfactuals("cf")
final_cols, win_cols = [f"final {x}" for x in LABELS], [f"win {x}" for x in LABELS]
W, T = cf[win_cols].to_numpy(), cf[final_cols].to_numpy()
best = (
    pd.Series(T.argmax(1)).map(dict(enumerate(LABELS))).value_counts(normalize=True).reindex(LABELS).fillna(0)
)
display(
    pd.DataFrame(
        {"win rate": W.mean(0), "best intent in (share of states)": best.to_numpy()}, index=LABELS
    ).round(3)
)
same = (cf["final withdraw/focus"] - cf["final withdraw/nearest"]).abs().max()
print(
    f"{len(cf)} states. The intent decides the winner in {(W.max(1) != W.min(1)).mean():.0%} of them; "
    f"best fixed intent wins {W.mean(0).max():.0%}, best of six in hindsight {W.max(1).mean():.0%}."
)
print(f"withdraw/focus and withdraw/nearest are the same behaviour: largest difference in outcome {same:.1e}")
```

## D6 · The planner in the real game (finding 9)

400 close battles, the same seeds for every policy (`scripts/eval_planners.py`).

```python
plan = pd.read_csv(PROCESSED / "planner_main.csv")
names = {
    "fixed:hold/nearest": "always hold/nearest",
    "random switching": "random switching",
    "fixed:advance/nearest": "always advance/nearest",
    "fixed:advance/focus": "always advance/focus",
    "model:sparse:model:4:12": "planner with the SINDYc model",
    "oracle": "planner with the real game (oracle)",
}
plan = plan[plan["policy"].isin(names)].assign(policy=lambda t: t["policy"].map(names))
won = plan.pivot(index="seed", columns="policy", values="won").astype(float)[list(names.values())]
shares = plan.fillna(0).groupby("policy")[[c for c in plan.columns if c.startswith("share_")]].mean()
shares.columns = [c.removeprefix("share_") for c in shares.columns]
summary = pd.DataFrame({"Blue wins": won.mean(), "std error": won.std() / np.sqrt(len(won))}).join(
    shares.round(2)
)
display(summary.round(3))
gap = (won["planner with the SINDYc model"] - won["always advance/focus"]).to_numpy()
boot = np.random.default_rng(0).choice(gap, (5000, len(gap))).mean(1)
print(
    f"SINDYc planner - always advance/focus: {gap.mean():+.3f}, 95% interval "
    f"[{np.quantile(boot, 0.025):+.3f}, {np.quantile(boot, 0.975):+.3f}]"
)
```

## D7 · Ranking intents: the model vs learning the choice directly (finding 10)

On the `cf` states: pick the intent with the best predicted final margin. The SINDYc model rolls
each intent out for 16 s. The direct alternative learns "withdraw - advance" and
"focus - nearest" from the 1,803 `cftrain` states (team features only; notebook 3 shows unit
features add nothing).

```python
def score(choice):
    idx = np.arange(len(cf))
    return {"win rate": W[idx, choice].mean(), "regret (final margin)": (T.max(1) - T[idx, choice]).mean()}


def predicted_margins(model, start, horizon=128):
    X0, S0, R0 = (start[c].to_numpy(float) for c in (STATE, SHAPE, REACH_COLS))
    out = []
    for i in INTENTS:
        U = np.repeat(np.array([1.0, 0, 0, 0])[None], horizon, 0)  # advance/nearest after the branch
        U[:32] = [i[0] == "advance", i[0] == "hold", i[0] == "withdraw", i[1] == "focus"]
        Y = model.simulate(X0, S0, np.repeat(U[None], len(X0), 0), reach0=R0, shape="model")
        out.append(Y[:, -1, 0] - Y[:, -1, 1])
    return np.column_stack(out)


cft, cft_start = load_counterfactuals("cftrain")
for start in (cf_start, cft_start):
    start["ratio"] = start["health_blue"] / start["health_red"]
team = STATE + SHAPE + REACH_COLS + ["ratio"]


def learned(a, b):
    y = (cft[f"final {a}"] - cft[f"final {b}"]).to_numpy()
    model = HistGradientBoostingRegressor(max_iter=200, max_depth=3, learning_rate=0.05)
    cv = cross_val_predict(model, cft_start[team], y, cv=5)
    return model.fit(cft_start[team], y).predict(cf_start[team]), r2(y, cv)


wd, r2_wd = learned("withdraw/nearest", "advance/nearest")
fo, _ = learned("advance/focus", "advance/nearest")
options = np.array([LABELS.index(x) for x in ("advance/nearest", "advance/focus", "withdraw/nearest")])
choices = {
    "always advance/nearest": np.full(len(cf), 0),
    "always advance/focus": np.full(len(cf), 1),
    "SINDYc model picks": predicted_margins(sparse, cf_start).argmax(1),
    "learned from cftrain picks": options[np.column_stack([np.zeros(len(cf)), fo, wd]).argmax(1)],
    "hindsight best": T.argmax(1),
}
print(f"withdraw - advance, cross-validated R² from the team state: {r2_wd:.2f}")
pd.DataFrame({k: score(v) for k, v in choices.items()}).T.round(3)
```
