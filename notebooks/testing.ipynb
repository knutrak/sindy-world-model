```python
import numpy as np

from sindy_wm.envs.spawn_sampling import sample_opposing_spawns

spawn = sample_opposing_spawns(
    np.random.default_rng(1), bounds=(0, 31, 8, 23), min_distance=8, max_distance=16
)
spawn.ally_pos, spawn.enemy_pos, spawn.attack_point
```

```python
from sindy_wm.envs.episode import play_episode
from sindy_wm.envs.policies import AttackNearestPolicy
from sindy_wm.envs.scenarios import make_env
from sindy_wm.evaluation.plots import plot_level_a
from sindy_wm.states.aggregate import level_a

env = make_env("5m_vs_5m", step_mul=2, spawn=spawn)  # built-in name or file in configs/scenarios/
episode = play_episode(env, AttackNearestPolicy(aggression=0.8), seed=1)

states = level_a(episode.snapshots)  # one row per step
lvl_ploy = plot_level_a(states, title="5m_vs_5m, seed 0")
```
